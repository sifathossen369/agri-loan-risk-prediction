```python
climate_features = [
    'Total_Rainfall_(mm)',
    'Cloud Coverage (Octs)',
    'Humidity',
    'Maximum_Temperature',
    'Minimum_Temperature',
    'Sunshine',
    'Windspeed',
    'Eto',
    'PET',
    'Pressure',
    'WindDir',
    'WindGustSpeed',
    'WindGustDir',
    'weather_stress_index'
]
```

```python
X_A = df[financial_features]

y_A = df["loan_default"]
```

```python
print(X_A.shape)
print(y_A.shape)
```

```python
X_A.dtypes
```

```python


X_train_A, X_test_A, y_train_A, y_test_A = train_test_split(
    X_A,
    y_A,
    test_size=0.2,
    random_state=42,
    stratify=y_A
)
```

```python
print("Train:", X_train_A.shape)
print("Test:", X_test_A.shape)
```

```python
from sklearn.linear_model import LogisticRegression

lr_A = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_A.fit(X_train_A, y_train_A)
```

```python
y_pred_A = lr_A.predict(X_test_A)
```

```python
y_prob_A = lr_A.predict_proba(X_test_A)[:, 1]
```

```python
from sklearn.metrics import accuracy_score, precision_score
from sklearn.metrics import recall_score, f1_score, roc_auc_score

print("Accuracy:", accuracy_score(y_test_A, y_pred_A))
print("Precision:", precision_score(y_test_A, y_pred_A))
print("Recall:", recall_score(y_test_A, y_pred_A))
print("F1 Score:", f1_score(y_test_A, y_pred_A))
print("ROC-AUC:", roc_auc_score(y_test_A, y_prob_A))
```

```python
from sklearn.metrics import confusion_matrix

cm_A = confusion_matrix(y_test_A, y_pred_A)

print(cm_A)
```

```python
results = []

results.append({
    "Approach": "A - Financial Only",
    "Model": "Logistic Regression",
    "Accuracy": accuracy_score(y_test_A, y_pred_A),
    "Precision": precision_score(y_test_A, y_pred_A),
    "Recall": recall_score(y_test_A, y_pred_A),
    "F1": f1_score(y_test_A, y_pred_A),
    "ROC_AUC": roc_auc_score(y_test_A, y_prob_A)
})
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_A = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

rf_A.fit(X_train_A, y_train_A)
```
